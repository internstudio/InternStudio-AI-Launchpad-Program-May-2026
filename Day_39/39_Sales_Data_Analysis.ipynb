### 1. Total Sales Revenue

Calculate the total revenue generated from all sales transactions.

```python
total_sales = df['Sales_Amount'].sum()
print("Total Sales Revenue is :",total_sales)
```

### 2. Total Quantity Sold

Find the total number of products sold across all transactions.

```python
total_products = df['Quantity_Sold'].sum()
print("Total Quantity Sold is :",total_products)
```

### 3. Average Sales Amount

Calculate the average value of each sales transaction.

```python
total_sales_avg = df['Sales_Amount'].mean()
print("Average Sales Revenue is :",total_sales_avg)
```

### 4. Highest Sales Transaction

Identify the transaction with the highest sales amount.

```python
total_sales_max = df['Sales_Amount'].max()
print("Highest sales transaction is :",total_sales_max)
```

### 5. Lowest Sales Transaction

Identify the transaction with the lowest sales amount.

```python
total_sales_min = df['Sales_Amount'].min()
print("Lowest sales transaction is :",total_sales_min)
```

### 15. Discount Analysis

Analyze the average, minimum, and maximum discounts offered across all transactions.

```python
discount_average = df['Discount'].mean()
discount_minimum = df['Discount'].min()
dicount_maximum = df['Discount'].max();
print("Average Discount is :",discount_average * 100)
print("Minimum Discount is :",discount_minimum * 100)
print("Maximum Discount is :",dicount_maximum * 100)
```

### 16. Unit Cost vs Unit Price Analysis

Compare the average cost price and selling price of products to understand pricing strategies.

### 17. Profit Analysis

Calculate the profit earned on each transaction and identify overall business profitability.

```python
df['Profit'] = [(df['Unit_Price'][i] - df['Unit_Cost'][i]) * df['Quantity_Sold'][i] for i in range(len(df))]
```

### 18. Profit by Region

### 19. Monthly Sales Trend Analysis

### 25. Business Insights Report

### 1. Bar Chart – Region-wise Sales Analysis

**Objective**: Compare the total sales generated by each region.

```python
region_sales = df.groupby("Region")["Sales_Amount"].sum()

plt.figure(figsize=(8,5))
plt.bar(region_sales.index, region_sales.values)

plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

### 2. Bar Chart – Sales by Product Category

**Objective**: Compare sales across different product categories.

```python
category_sales = df.groupby("Product_Category")["Sales_Amount"].sum()

plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Sales Amount")
plt.show()
```

### 3. Line Chart – Monthly Sales Trend

**Objective**: Analyze how sales change over different months.

```python
df["Sale_Date"] = pd.to_datetime(df["Sale_Date"])
df["Month"] = df["Sale_Date"].dt.month

monthly_sales = df.groupby("Month")["Sales_Amount"].sum()

plt.figure(figsize=(8,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

### 4. Pie Chart – Customer Type Distribution

**Objective**: Visualize the percentage of New and Returning customers.

```python
customer = df["Customer_Type"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(customer.values,
        labels=customer.index,
        autopct="%1.1f%%")

plt.title("Customer Type Distribution")
plt.show()
```

### 5. Pie Chart – Payment Method Distribution

**Objective**: Show how customers prefer to make payments.

```python
payment = df["Payment_Method"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(payment.values,
        labels=payment.index,
        autopct="%1.1f%%")

plt.title("Payment Method Distribution")
plt.show()
```

### 6. Histogram – Sales Amount Distribution

**Objective**: Understand the distribution of sales amounts.

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales_Amount"], bins=20)

plt.title("Sales Amount Distribution")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.show()
```

### 7. Scatter Plot – Quantity Sold vs Sales Amount

**Objective**: Analyze the relationship between quantity sold and sales amount.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Quantity_Sold"], df["Sales_Amount"])

plt.title("Quantity Sold vs Sales Amount")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales Amount")
plt.show()
```

### 8. Scatter Plot – Unit Cost vs Unit Price

**Objective**: Compare the relationship between product cost and selling price.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Unit_Cost"], df["Unit_Price"])

plt.title("Unit Cost vs Unit Price")
plt.xlabel("Unit Cost")
plt.ylabel("Unit Price")
plt.show()
```

### 9. Bar Chart – Sales by Sales Representative

**Objective**: Compare the performance of different sales representatives.

```python
sales_rep = df.groupby("Sales_Rep")["Sales_Amount"].sum()

plt.figure(figsize=(8,5))
plt.bar(sales_rep.index, sales_rep.values)

plt.title("Sales by Sales Representative")
plt.xlabel("Sales Representative")
plt.ylabel("Sales Amount")
plt.show()
```

### 10. Line Chart – Daily Sales Trend

**Objective**: Track sales performance on a daily basis.

```python
daily_sales = df.groupby("Sale_Date")["Sales_Amount"].sum()

plt.figure(figsize=(12,5))
plt.plot(daily_sales.index, daily_sales.values)

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```
