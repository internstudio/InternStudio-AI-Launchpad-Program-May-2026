Writing data to the file with multiple lines

```python
file = open("employee.txt","w")

file.writelines(['Welcome to the file processing 1!\n', 'Welcome to the file processing 2!\n', 'Welcome to the file processing 3!\n', 'Welcome to the file processing 4!\n', 'Welcome to the file processing 5!\n', 'Welcome to the file processing 6!'])

file.close()
```

Appending content to the file

```python
file  = open("employee.txt","a")

file.write("\nThis is the new line we are appending")

file.close()
```

Creating a new file

```python
file = open("employee1.txt","x")

file.write("Hello")

file.close()
```

```python
with open("employee1.txt","r") as f1:
    print(f1.read())
```

Reading a file using loops

```python
with open("employee.txt") as file:
    for line in file:
        print(line)
```

```python
with open("example.txt","r+") as file:
    content = file.read()
    print(content)

    # insert new content to the file
    file.write("New content is here")
```

```python
with open("example.txt","w+") as file:
    content = file.read()
    print(content)

    # insert new content to the file
    file.write("New content is here")
```

```python
with open("example.txt","a+") as file:
    content = file.read()
    print(content)

    # insert new content to the file
    file.write("New content is here")
```
