```python
import numpy as np
```

```python
arr1 = np.array([1,2,38,9,8,8,6,85,0,12,34,4565,67,7])
arr2 = np.array([1,2,34,9,8,7,6,5,0,12,34,4565,67,7])
```

Addtition of two arrays

```python
result2 = arr1 + arr2
print(result2)
```

Performing substraction on arrays

```python
result = arr1 - arr2
print(result)
result2 = np.subtract(arr1,arr2)
print(result2)
```

Multiplicaiton

```python
result = arr1 * arr2
print(result)

result = np.multiply(arr1,arr2)
print(result)
```

Devision of an array

```python
result = arr1 / arr2
print(result)

result = np.divide(arr1,arr2)
print(result)
```

Applying power to the array

```python
result = np.power(arr1,2)
print(result)
```

```python
result = np.power(arr1,arr2)
print(result)
```

Modulos of array

```python
result = arr1 % 4
print(result)

result = np.mod(arr1,4)
print(result)
```

Scalar Operations

```python
arr = np.array([10,20,30])
print(arr + 5)

print(arr * 2)

print(arr / 2)

print(arr ** 2)
```

Comparison Operations

```python
arr = np.array([10,20,30])

result = arr == 20
print(result)

result = arr != 20
print(result)

result = arr > 20
print(result)

result = arr < 20
print(result)
```

```python
a = np.array([True, False, True])
b = np.array([True, True, False])

print("Logical AND",np.logical_and(a,b))
print("Logical OR",np.logical_or(a,b))

print("Logical NOT",np.logical_not(a))
```

Aggragator Operations

```python
print("Sum : ",arr1.sum())
print("Min : ",arr1.min())
print("Max : ",arr1.max())
print("Mean : ",arr1.mean())
print("Median : ",np.median(arr1))
print("Variance : ",np.var(arr1))
print("Standard Deviation : ",np.std(arr1))
print("Correlation Coefficient : ",np.corrcoef(arr1))
print("Product",np.prod(arr1))
print("Cummulative Sum",np.cumsum(arr1))
print("Cummulative Product",np.cumprod(arr1))
```

Absolute Values

```python
arr = np.array([-10,-20,15])
print(np.abs(arr))
```

```python
arr = np.array([2.34,4.78,6.12])
# Round -> prev | next
print(np.round(arr))

# Floor -> prev
print(np.floor(arr))

# Ceiling -> next
print(np.ceil(arr))
```

Matrix Operaitons

```python
a = np.array([
 [1,2],
 [3,4]
])
b = np.array([
 [5,6],
 [7,8]
])
print(a + b)
print(a * b)
```
