Create an array with numpy

```python
arr = np.array([10,12,23,34,45,21,32,43,54,20,30,40,50])
print(arr)
```

create an array with list

```python
lst = [1,2,3,4,56,7,78,9,90]
arr2 = np.array(lst)
```

```python
print(arr2)
```

Crate an array with tuple

```python
data = (12,34,5,6,6,43,7,78,8)
arr3 = np.array(data)
print(arr3)
```

One-Dimensional Array (1D Array)
A 1D array contains only one row

```python
arr = np.array([10,20,30,40])
print(arr)
```

Two-Dimensional Array (2D Array)

```python
twoDarr = np.array([[1,2,34,5,6,7],[1,2,3,4,5,6]])
```

```python
print(twoDarr)
```

Three Dimensional Array

```python
arr = np.array([
 [
 [1,2,3],
 [3,4,3]
 ],
 [
 [5,6,3],
 [7,83,3]
 ]
])
```

```python
print(arr)
```

checking the data type of an array
