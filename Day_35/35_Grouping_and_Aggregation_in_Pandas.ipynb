```python
import pandas as pd
```

```python
data = {
 "Employee":["Amit","Rahul","Priya","Neha","Karan","Anjali","Rohit","Sneha"],
 "Department":["IT","HR","IT","HR","Sales","Sales","IT","HR"],
 "City":["Mumbai","Delhi","Mumbai","Delhi","Pune","Pune","Delhi","Mumbai"],
 "Salary":[70000,50000,80000,55000,45000,48000,75000,60000],
 "Experience":[5,2,6,3,2,4,5,7]
}
```

```python
print(data)
```

Multiple Aggregations
df.groupby("Department")["Salary"].agg(["sum","mean","min","max"])

```python
df.groupby("Department")["Salary"].agg(["min","max","mean","sum","var"])
```

Custom Column Names
df.groupby("Department").agg(
 Total_Salary=("Salary","sum"),
 Average_Salary=("Salary","mean"),
 Highest_Salary=("Salary","max")
)

```python
df.groupby("Department").agg(
    Minimum_Salary=("Salary","min"),
    Maximum_Salary=("Salary","max"),
    Average_Salary=("Salary","mean"),
    Total_Salary=("Salary","sum")
)
```

```python
df.groupby("Department").agg(
    Minimum_Experience=("Experience","min"),
    Maximum_Experience=("Experience","max"),
    Average_Experience=("Experience","mean"),
    Total_Experience=("Experience","sum")
)
```

Groupoing Multiple Columns

```python
df.groupby(["Department","City"])["Salary"].sum()
```

```python
df.groupby(["Department","City"])["Salary"].agg(["sum","min"])
```

Selecting Multiple Columns
df.groupby("Department")[["Salary","Experience"]].mean()

```python
df.groupby("City")[["Salary","Experience"]].mean()
```

```python
df.groupby("City")[["Salary","Experience"]].agg(["sum","min","max"])
```

Filtering Groups

df.groupby("Department").filter(
 lambda x: x["Salary"].mean() > 50000
)

```python
df.groupby("City").filter(lambda x: x["Salary"].min() > 50000)
```

```python
df.groupby("City").filter(lambda x: x["Experience"].min() > 3)
```

Transform

```python
df["Average Salary"] = df.groupby("Department")["Salary"].transform("mean")
```

```python
df.head()
```

```python
df["Min Salary"] = df.groupby("Department")["Salary"].transform("min")
```

```python
df.head()
```

Sorting Grouped Results

```python
df.groupby("Department")["Salary"].sum().sort_values(ascending=True)
```
