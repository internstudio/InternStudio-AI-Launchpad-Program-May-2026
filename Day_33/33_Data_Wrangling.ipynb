```python
import numpy as np

# Identify numeric columns
numeric_cols = df.select_dtypes(include=np.number).columns

# Get 50 random indices
num_rows = len(df)
random_indices = np.random.choice(num_rows, 50, replace=False)

# Introduce 50 null entries into numeric columns
for col in numeric_cols:
    # For simplicity, we'll assign NaN to 50 random rows in each numeric column
    # A more precise way would be to ensure exactly 50 total nulls across all numeric columns combined
    # but for adding '50 null entries in the columns which are numeric', this is a direct interpretation.
    df.loc[random_indices, col] = np.nan

print("Null entries added to numeric columns.")
```

```python
# Display the updated null counts
df.isnull().sum()
```
