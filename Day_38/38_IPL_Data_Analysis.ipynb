## IPL Data Analysis Tasks

```python
# 1. Dataset Shape
print(f"Dataset Shape: {df.shape}")
```

```python
# 2. Column Names
print("Column Names:")
for col in df.columns:
    print(f"- {col}")
```

```python
# 3. Data Types
print("Data Types:")
display(df.dtypes)
```

```python
# 4. Summary Statistics
print("Summary Statistics:")
display(df.describe())
```

```python
# 5. Total Matches
total_matches = df["match_id"].nunique()
print(f"Total Matches: {total_matches}")
```

```python
# 6. Total Players
total_players = df["batter"].nunique()
print(f"Total Players: {total_players}")
```

```python
# 7. Total Teams
total_teams = df["batting_team"].nunique()
print(f"Total Teams: {total_teams}")
```

```python
# 8. Total Venues
total_venues = df["venue"].nunique()
print(f"Total Venues: {total_venues}")
```

```python
# 9. Total Cities
total_cities = df["city"].nunique()
print(f"Total Cities: {total_cities}")
```

```python
# 10. Total Runs Scored
total_runs_scored = df["runs_total"].sum()
print(f"Total Runs Scored: {total_runs_scored}")
```

```python
# 11. Total Boundaries (balls where runs_batter is 4 or 6)
boundaries = df[df["runs_batter"].isin([4,6])]
print(f"Total Boundaries (4s and 6s): {len(boundaries)}")
```

```python
# 12. Total Fours
total_fours = (df["runs_batter"]==4).sum()
print(f"Total Fours: {total_fours}")
```

```python
# 13. Total Sixes
total_sixes = (df["runs_batter"]==6).sum()
print(f"Total Sixes: {total_sixes}")
```

```python
# 14. Average Runs Per Ball
average_runs_per_ball = df["runs_total"].mean()
print(f"Average Runs Per Ball: {average_runs_per_ball:.2f}")
```

```python
# 15. Highest Runs in One Ball
highest_runs_one_ball = df["runs_total"].max()
print(f"Highest Runs in One Ball: {highest_runs_one_ball}")
```

```python
# 16. Top 10 Batters (by runs scored)
top_10_batters = df.groupby("batter")["runs_batter"].sum().sort_values(ascending=False).head(10)
print("Top 10 Batters:")
display(top_10_batters)
```

```python
# 17. Bottom 10 Batters (by runs scored)
bottom_10_batters = df.groupby("batter")["runs_batter"].sum().sort_values().head(10)
print("Bottom 10 Batters:")
display(bottom_10_batters)
```

```python
# 18. Top 10 Bowlers (by wickets taken)
top_10_bowlers = df.groupby("bowler")["bowler_wicket"].sum().sort_values(ascending=False).head(10)
print("Top 10 Bowlers:")
display(top_10_bowlers)
```

```python
# 19. Most Dot Balls (runs_total == 0)
dot_balls_bowler = df[df["runs_total"]==0].groupby("bowler").size().sort_values(ascending=False).head(10)
print("Bowlers with Most Dot Balls:")
display(dot_balls_bowler)
```

```python
# 20. Most Sixes by Batter
most_sixes_batter = df[df["runs_batter"]==6].groupby("batter").size().sort_values(ascending=False).head(10)
print("Batters with Most Sixes:")
display(most_sixes_batter)
```

```python
# 21. Most Fours by Batter
most_fours_batter = df[df["runs_batter"]==4].groupby("batter").size().sort_values(ascending=False).head(10)
print("Batters with Most Fours:")
display(most_fours_batter)
```

```python
# 22. Strike Rate Calculation
strike_rate_stats = df.groupby("batter").agg(
    Runs=("runs_batter","sum"),
    Balls_Faced=("valid_ball","sum")
)
strike_rate_stats["Strike Rate"] = (strike_rate_stats["Runs"] / strike_rate_stats["Balls_Faced"]) * 100
print("Batter Strike Rates:")
display(strike_rate_stats.sort_values(by="Strike Rate", ascending=False).head(10))
```

```python
# 23. Economy Rate Calculation
economy_rate_stats = df.groupby("bowler").agg(
    Runs_Conceded=("runs_bowler","sum"),
    Balls_Bowled=("valid_ball","sum")
)
economy_rate_stats["Overs_Bowled"] = economy_rate_stats["Balls_Bowled"] / 6
# Avoid division by zero for bowlers who bowled 0 overs
economy_rate_stats["Economy_Rate"] = economy_rate_stats.apply(
    lambda row: row["Runs_Conceded"] / row["Overs_Bowled"] if row["Overs_Bowled"] > 0 else 0,
    axis=1
)
print("Bowler Economy Rates:")
display(economy_rate_stats.sort_values(by="Economy_Rate").head(10))
```

```python
# 24. Highest Individual Score
highest_individual_score = df.groupby(["match_id","batter"])["runs_batter"].sum().sort_values(ascending=False).head(10)
print("Highest Individual Scores:")
display(highest_individual_score)
```

```python
# 25. Highest Team Score
highest_team_score = df.groupby(["match_id","batting_team"])["team_runs"].max().sort_values(ascending=False).head(10)
print("Highest Team Scores:")
display(highest_team_score)
```

```python
# 26. Runs by Team
runs_by_team = df.groupby("batting_team")["runs_total"].sum().sort_values(ascending=False)
print("Total Runs Scored by Each Team:")
display(runs_by_team)
```

```python
# 27. Wickets by Team
wickets_by_team = df.groupby("bowling_team")["bowler_wicket"].sum().sort_values(ascending=False)
print("Total Wickets Taken by Each Team:")
display(wickets_by_team)
```

```python
# 28. Matches Played by Team
matches_played_by_team = df.groupby("batting_team")["match_id"].nunique().sort_values(ascending=False)
print("Number of Matches Played by Each Team:")
display(matches_played_by_team)
```

```python
# 29. Matches Won by Each Team
matches_won = df["match_won_by"].value_counts()
print("Matches Won by Each Team:")
display(matches_won)
```

```python
# 30. Toss Winners
toss_winners = df["toss_winner"].value_counts()
print("Toss Winners:")
display(toss_winners)
```

```python
# 31. Toss Decision Analysis
toss_decision_analysis = df["toss_decision"].value_counts()
print("Toss Decision Analysis:")
display(toss_decision_analysis)
```

```python
# 32. Player of Match Awards
player_of_match_awards = df["player_of_match"].value_counts().head(10)
print("Top 10 Player of Match Award Winners:")
display(player_of_match_awards)
```

```python
# 33. Venue Wise Matches
venue_wise_matches = df.groupby("venue")["match_id"].nunique().sort_values(ascending=False)
print("Matches Played at Each Venue:")
display(venue_wise_matches.head(10))
```

```python
# 34. City Wise Matches
city_wise_matches = df["city"].value_counts()
print("Matches Played in Each City:")
display(city_wise_matches.head(10))
```

```python
# 35. Runs by Venue
runs_by_venue = df.groupby("venue")["runs_total"].sum().sort_values(ascending=False)
print("Total Runs Scored at Each Venue:")
display(runs_by_venue.head(10))
```

```python
# 36. Runs by Season
runs_by_season = df.groupby("season")["runs_total"].sum().sort_values(ascending=False)
print("Total Runs Scored Per Season:")
display(runs_by_season)
```

```python
# 37. Matches by Season
matches_by_season = df.groupby("season")["match_id"].nunique().sort_values(ascending=False)
print("Number of Matches Per Season:")
display(matches_by_season)
```

```python
# 38. Runs by Year
runs_by_year = df.groupby("year")["runs_total"].sum().sort_values(ascending=False)
print("Total Runs Scored Per Year:")
display(runs_by_year)
```

```python
# 39. Matches by Year
matches_by_year = df.groupby("year")["match_id"].nunique().sort_values(ascending=False)
print("Number of Matches Per Year:")
display(matches_by_year)
```

```python
# 40. Wicket Types
wicket_types = df["wicket_kind"].value_counts()
print("Distribution of Wicket Types:")
display(wicket_types)
```

```python
# 41. Most Common Extras
most_common_extras = df["extra_type"].value_counts()
print("Most Common Extra Types:")
display(most_common_extras)
```

```python
# 42. Powerplay Runs (overs 1-6)
powerplay_runs = df[df["over"] <= 6]["runs_total"].sum()
print(f"Total Runs Scored in Powerplay (Overs 1-6): {powerplay_runs}")
```

```python
# 43. Death Overs Runs (overs 16-20)
death_overs_runs = df[df["over"] >= 16]["runs_total"].sum()
print(f"Total Runs Scored in Death Overs (Overs 16-20): {death_overs_runs}")
```

```python
# 44. Average Team Score Per Match
average_team_score_per_match = df.groupby(["match_id","batting_team"])["team_runs"].max().mean()
print(f"Average Team Score Per Match: {average_team_score_per_match:.2f}")
```

```python
# 45. Average Wickets Per Match
average_wickets_per_match = df.groupby("match_id")["bowler_wicket"].sum().mean()
print(f"Average Wickets Per Match: {average_wickets_per_match:.2f}")
```

```python
# 46. Distribution of Runs Per Ball
plt.figure(figsize=(10, 6))
plt.hist(df["runs_total"], bins=range(df["runs_total"].min(), df["runs_total"].max() + 2), align='left', rwidth=0.8)
plt.title("Distribution of Runs Per Ball")
plt.xlabel("Runs Scored on a Ball")
plt.ylabel("Frequency")
plt.xticks(range(df["runs_total"].min(), df["runs_total"].max() + 1))
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# 47. Top 10 Partnerships (based on total runs scored together)
# Note: 'batting_partners' column needs to be cleaned/parsed if it's a string representation of a tuple
# Assuming 'batting_partners' is already in a usable format or can be grouped as is.
# If 'batting_partners' is a string like "('A', 'B')", it needs to be processed to be consistent.
# For now, I'll group as is, which might treat different orders as different partnerships if not preprocessed.
# A more robust solution might involve sorting the names within the tuple string.
top_10_partnerships = df.groupby("batting_partners")["runs_total"].sum().sort_values(ascending=False).head(10)
print("Top 10 Batting Partnerships (by total runs):")
display(top_10_partnerships)
```

```python
# 48. Correlation Analysis for selected columns
selected_cols_for_corr = [
    "runs_batter",
    "balls_faced",
    "runs_total",
    "team_runs",
    "team_wicket"
]
correlation_matrix = df[selected_cols_for_corr].corr()
print("Correlation Matrix for Selected Variables:")
display(correlation_matrix)
```

```python
# 49. Highest Scoring Overs
highest_scoring_overs = df.groupby("over")["runs_total"].sum().sort_values(ascending=False)
print("Highest Scoring Overs (by total runs):")
display(highest_scoring_overs)
```

```python
# 50. Run Distribution by Batter (0, 1, 2, 3, 4, 5, 6)
run_distribution_batter = df["runs_batter"].value_counts().sort_index()
print("Distribution of Runs Scored by Batters (per ball):")
display(run_distribution_batter)
```

```python
# Fours vs Sixes by Batter (Stacked Bar Chart)
fours_sixes = df[df['runs_batter'].isin([4, 6])].groupby('batter')['runs_batter'].value_counts().unstack(fill_value=0)
fours_sixes['Total Boundaries'] = fours_sixes[4] + fours_sixes[6]
top_boundary_hitters = fours_sixes.sort_values(by='Total Boundaries', ascending=False).head(10)

plt.figure(figsize=(14, 7))
top_boundary_hitters[[4, 6]].plot(kind='bar', stacked=True, colormap='viridis')
plt.title('Top 10 Batters: Fours vs Sixes')
plt.xlabel('Batter')
plt.ylabel('Number of Boundaries')
plt.xticks(rotation=45, ha='right')
plt.legend(['Fours', 'Sixes'])
plt.tight_layout()
plt.show()
```

```python
# Average Team Scores by Season (Line Chart)
average_team_score_by_season = df.groupby(['season', 'batting_team'])['team_runs'].max().groupby('season').mean()

plt.figure(figsize=(12, 6))
sns.lineplot(x=average_team_score_by_season.index.astype(str), y=average_team_score_by_season.values, marker='o')
plt.title('Average Team Score Per Season')
plt.xlabel('Season')
plt.ylabel('Average Team Score')
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Powerplay vs Death Overs Runs (Bar Chart)
runs_comparison = pd.DataFrame({
    'Over Type': ['Powerplay (Overs 1-6)', 'Death Overs (Overs 16-20)'],
    'Total Runs': [powerplay_runs, death_overs_runs]
})

plt.figure(figsize=(8, 6))
sns.barplot(x='Over Type', y='Total Runs', data=runs_comparison, palette='coolwarm')
plt.title('Total Runs Scored in Powerplay vs Death Overs')
plt.xlabel('Over Type')
plt.ylabel('Total Runs')
plt.tight_layout()
plt.show()
```
