Handling Specific Exceptions
It is recommended to handle specific exceptions rather than using a generic except block.

```python
ZeroDivisionError
try:
 num = 10 / 0
except ZeroDivisionError:
 print("Cannot divide by zero.")
```

```python
# ValueError
# Occurs when the user enters an invalid value.
try:
 age = int(input("Enter age: "))
except ValueError:
 print("Please enter a valid number.")
```

```python
# NameError
try:
 print(student_name)
except NameError:
 print("Variable not defined.")
```

```python
# TypeError
try:
 result = "10" + 5
except TypeError:
 print("Cannot combine string and integer.")
```

```python
FileNotFoundError
try:
 file = open("data.txt", "r")
except FileNotFoundError:
 print("File does not exist.")
```

```python
try:
    num = int(input("Enter the number: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError:
    print("Value error occured")
except:
    print("Unknown exception occured")
```

Using else Block
The else block executes only if no exception occurs.
Syntax
try:
 pass
except:
 pass
else:
 pass

```python
try:
 num = int(input("Enter a number: "))
except ValueError:
 print("Invalid Input")
else:
 print("you entered",num)
```

finally syntax

try:
    pass
except:
    pass
finally:
    pass

```python
try:
    file = open("data.txt","r")
    print(file.read())
except FileNotFoundError as e:
    print(e)
finally:
    file.close()
    print("Program execution completed")
```

```python
try:
 num1 = int(input("Enter first number: "))
 num2 = int(input("Enter second number: "))
 result = num1 / num2
except ValueError:
 print("Please enter valid numbers.")
except ZeroDivisionError:
 print("Division by zero is not allowed.")
else:
 print("Result:", result)
finally:
 print("Program execution finished.")
```
