```python
import numpy as np

# Get the number of rows and columns
num_rows, num_cols = df.shape

# Generate 50 unique random row and column indices
np.random.seed(42) # for reproducibility
random_indices = set()
while len(random_indices) < 50:
    row_idx = np.random.randint(0, num_rows)
    col_idx = np.random.randint(0, num_cols)
    random_indices.add((row_idx, col_idx))

# Convert to list of tuples for easier iteration
random_indices_list = list(random_indices)

# Add null entries
for r, c in random_indices_list:
    # Using iloc for integer-location based indexing
    df.iloc[r, c] = pd.NA

print(f"Successfully added {len(random_indices_list)} null entries to the DataFrame.")
```
