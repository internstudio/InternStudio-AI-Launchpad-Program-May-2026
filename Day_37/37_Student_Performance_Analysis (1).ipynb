Students got grade A
Students got grade B

grade >= 40 : A
grade > 40 and grade < 30 : B  
grade >= 30 and grade < 20 C
grade > 20 and grade < 10 : D

```python
df['grade'] = df['exam_score'].apply(lambda x : 'A' if x >= 40 else ('B' if x < 40 and x > 30 else ('C' if x < 30 and x < 20 else 'D')))
```

```python
df.groupby('grade')['exam_score'].agg(["min","max","mean","sum"])
```
