```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales = [120,145,180,170,210]
```

```python
plt.plot(months,sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
```

```python
# Changing color
plt.plot(months,sales,color="red")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
```

Styles:

```python
from matplotlib.lines import lineStyles
plt.plot(months,sales,linestyle=":")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
```

```python
plt.plot(months,sales,marker="^")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
```

```python
# Example of all the parameters
months = ["Jan","Feb","Mar","Apr","May"]
sales = [120,145,180,170,210]
plt.plot(
 months,
 sales,
 color="blue",
 marker="o",
 linestyle="-"
 )
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# bar Chart
plt.bar(months,sales,color="purple")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
```

```python
countries = ['China', 'India', 'United States', 'Indonesia', 'Pakistan']
population = [1425000000, 1408000000, 334800000, 277500000, 240400000]
```

```python
# bar Chart
plt.bar(countries,population,color="purple")
plt.xlabel("Countries")
plt.ylabel("Population")
plt.title("Population Analysis")
plt.grid(True)
plt.show()
```

```python
# bar Chart
plt.barh(countries,population,color="purple")
plt.xlabel("Countries")
plt.ylabel("Population")
plt.title("Population Analysis")
plt.grid(True)
plt.show()
```

```python
# Pie chart
subjects = ["Math","Science","English"]
marks = [40,35,25]
plt.pie(
    marks,
    labels=subjects,
    # colors=["red","green","blue"],
    autopct="%1.1f%%"
)

plt.title("Marks Analysis")
plt.show()
```

```python
hours = [1,2,11,9,8,6]
marks = [45,55,60,72,82,90]
```

```python
# Scatter plot
plt.scatter(hours,marks)
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Hours vs Marks")
plt.grid(True)
plt.show()
```

```python
# histogram
ages = [20,22,25,21,24,26,28,24,25,22,23,24,102,-1,-2,300]
plt.hist(ages,bins=[20,25,30,35,40])
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.grid(True)
plt.show()
```

```python
# boxplot
plt.boxplot(ages)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.grid(True)
plt.show()
```

```python
# Area chart
months = ["Jan","Feb","Mar","Apr","May"]
sales = [120,145,180,170,210]
plt.fill_between(months, sales)
plt.show()
```

```python
# Multiplotting
months = ["Jan","Feb","Mar","Apr","May"]
product1 = [100,120,140,160,180]
product2 = [90,130,135,170,200]
product3 = [80,90,100,122,121]

plt.plot(months,product1,label="Laptop")
plt.plot(months,product2,label="Mobile")
plt.plot(months,product3,label="Tablet")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.savefig("sales_chart.png")
plt.legend()
```
